# Medical Robotics Mini-Lab: Surgical Tool Trajectory
Prototype first in Colab, then move the same idea to VS Code, Docker and ROS 2.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Define the scenario
The robot starts at the origin and moves toward a target detected from medical imaging.

```python
start = np.array([0.0, 0.1])
target = np.array([0.12, 0.08])
obstacle = np.array([0.06, 0.04])
obstacle_radius = 0.02
```

## 2. Plan a straight-line trajectory

```python
n_points = 50
x = np.linspace(start[0], target[0], n_points)
y = np.linspace(start[1], target[1], n_points)
trajectory = np.column_stack((x, y))
trajectory[:5]
```

## 3. Visualize the plan
The obstacle represents anatomy that should not be crossed.

```python
fig, ax = plt.subplots(figsize=(7,5))
ax.plot(trajectory[:,0], trajectory[:,1], linewidth=2, label="Planned trajectory")
ax.scatter(*start, s=90, label="Start")
ax.scatter(*target, s=90, label="Target")
ax.add_patch(plt.Circle(obstacle, obstacle_radius, alpha=0.25, label="Anatomical structure"))
ax.set_xlabel("x [m]")
ax.set_ylabel("y [m]")
ax.set_title("Surgical Tool Trajectory")
ax.axis("equal")
ax.grid(True)
ax.legend()
plt.show()
```
